### QuickSelect Algorithm Implementation

Here's the implementation of the QuickSelect algorithm, which is a selection algorithm to find the k-th smallest element in an unsorted list. It is related to the Quicksort sorting algorithm.

```python
import random

def quickselect(arr, k):
    """Finds the k-th smallest element in an unsorted array using the QuickSelect algorithm."""

    if not arr:
        return None

    if k < 1 or k > len(arr):
        raise IndexError("k is out of bounds for the array")

    # Convert k to 0-indexed
    k_index = k - 1

    def _quickselect_recursive(sub_arr):
        if len(sub_arr) == 1:
            return sub_arr[0]

        pivot = random.choice(sub_arr)
        low = [x for x in sub_arr if x < pivot]
        high = [x for x in sub_arr if x > pivot]
        pivots = [x for x in sub_arr if x == pivot]

        if k_index < len(low):
            return quickselect(low, k)
        elif k_index < len(low) + len(pivots):
            return pivots[0]
        else:
            # k_index is in the high part, adjust k_index relative to the high part
            return quickselect(high, k_index - len(low) - len(pivots) + 1)

    # Start the recursive call with the original array
    # Note: The recursive helper function should be adjusted to work with k_index directly
    # For simplicity, let's refactor quickselect to be recursive itself

    # Re-implementing for clarity and correctness of recursion with k_index
    def _partition_and_select(arr_segment, target_index):
        if len(arr_segment) == 1:
            return arr_segment[0]

        pivot_index = random.randint(0, len(arr_segment) - 1)
        pivot = arr_segment[pivot_index]

        less = []
        equal = []
        greater = []

        for x in arr_segment:
            if x < pivot:
                less.append(x)
            elif x == pivot:
                equal.append(x)
            else:
                greater.append(x)

        if target_index < len(less):
            return _partition_and_select(less, target_index)
        elif target_index < len(less) + len(equal):
            return equal[0]
        else:
            return _partition_and_select(greater, target_index - len(less) - len(equal))

    return _partition_and_select(arr, k_index)


```

### Testing the QuickSelect Algorithm

Now, let's create an array of 1000 random integers and test the `quickselect` function for the minimum (k=1), maximum (k=n), and median (k=n//2) values.

```python
import numpy as np

# Generate an array of 1000 random integers
n = 1000
random_array = [random.randint(1, 10000) for _ in range(n)]

print(f"Generated array of {n} random integers.")

# Test with k=1 (minimum)
k_min = 1
min_element = quickselect(random_array, k_min)
print(f"\n{k_min}-th smallest element (min): {min_element}")

# Verify with built-in min function
actual_min = min(random_array)
print(f"Actual minimum (for verification): {actual_min}")
assert min_element == actual_min, "Minimum element mismatch!"

# Test with k=n (maximum)
k_max = n
max_element = quickselect(random_array, k_max)
print(f"\n{k_max}-th smallest element (max): {max_element}")

# Verify with built-in max function
actual_max = max(random_array)
print(f"Actual maximum (for verification): {actual_max}")
assert max_element == actual_max, "Maximum element mismatch!"

# Test with k=n//2 (median for odd n, or lower-median for even n)
k_median = n // 2
median_element = quickselect(random_array, k_median)
print(f"\n{k_median}-th smallest element (lower-median for even n): {median_element}")

# Verify with numpy median (after sorting)
sorted_array = sorted(random_array)
actual_median_index = k_median - 1 # k-th smallest is at index k-1 in a 0-indexed sorted array
actual_median = sorted_array[actual_median_index]
print(f"Actual median (from sorted array, index {actual_median_index}): {actual_median}")
assert median_element == actual_median, "Median element mismatch!"

print("\nAll tests passed successfully!")
```
