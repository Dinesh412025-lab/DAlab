```python
import time
import random
import string

def naive_search(text, pattern):
    n, m = len(text), len(pattern)
    matches, comparisons = [], 0
    for i in range(n - m + 1):
        j = 0
        while j < m:
            comparisons += 1
            if text[i + j] != pattern[j]:
                break
            j += 1
        if j == m:
            matches.append(i)
    return matches, comparisons

def compute_lps(pattern):
    m = len(pattern)
    lps = [0] * m
    length, i = 0, 1
    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        elif length != 0:
            length = lps[length - 1]
        else:
            lps[i] = 0
            i += 1
    return lps

def kmp_search(text, pattern):
    n, m = len(text), len(pattern)
    lps = compute_lps(pattern)
    matches, comparisons = [], 0
    i = j = 0
    while i < n:
        comparisons += 1
        if pattern[j] == text[i]:
            i += 1; j += 1
        if j == m:
            matches.append(i - j)
            j = lps[j - 1]
        elif i < n and pattern[j] != text[i]:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1
    return matches, comparisons

def rabin_karp(text, pattern, q=101):
    n, m = len(text), len(pattern)
    d = 256
    h = pow(d, m - 1, q)
    p_hash = t_hash = 0
    matches, comparisons = [], 0

    for i in range(m):
        p_hash = (d * p_hash + ord(pattern[i])) % q
        t_hash = (d * t_hash + ord(text[i])) % q

    for s in range(n - m + 1):
        if p_hash == t_hash:
            for k in range(m):
                comparisons += 1
                if text[s + k] != pattern[k]:
                    break
            else:
                matches.append(s)

        if s < n - m:
            t_hash = (d * (t_hash - ord(text[s]) * h) + ord(text[s + m])) % q
            if t_hash < 0:
                t_hash += q
    return matches, comparisons

# --- Main Execution ---
text = 'AABAACAADAABAABA'
pattern = 'AABA'
print(f'Text: {text}')
print(f'Pattern: {pattern}')

m1, c1 = naive_search(text, pattern)
m2, c2 = kmp_search(text, pattern)
m3, c3 = rabin_karp(text, pattern)

print(f'\nNaive -> Matches at: {m1}, Comparisons: {c1}')
print(f'KMP -> Matches at: {m2}, Comparisons: {c2}')
print(f'RK -> Matches at: {m3}, Comparisons: {c3}')

# Performance comparison
text_large = ''.join(random.choices('ABCD', k=10000))
patterns = ['AB', 'ABCD', 'ABCDAB', 'ABCDABCD']

print(f'\n{"Pattern":>12} {"Naive":>10} {"KMP":>10} {"RK":>10}')
print('-' * 50)

for p in patterns:
    _, c1 = naive_search(text_large, p)
    _, c2 = kmp_search(text_large, p)
    _, c3 = rabin_karp(text_large, p)
    print(f'{p:>12} {c1:>10} {c2:>10} {c3:>10}')





```

The code above implements three different string matching algorithms: Naive, Knuth-Morris-Pratt (KMP), and Rabin-Karp. It first demonstrates their usage with a sample text and pattern, showing the found matches and the number of character comparisons made. Then, it proceeds to a performance comparison section where it generates a large random text and tests the algorithms with various patterns, printing the total comparisons for each.

```python
# Further testing with a loop
num_iterations = 10

print(f'\nRunning {num_iterations} iterations for average comparisons:')
print(f'\n{"Pattern":>12} {"Naive (Avg)":>15} {"KMP (Avg)":>13} {"RK (Avg)":>13}')
print('-' * 60)

for p in patterns:
    naive_total_comparisons = 0
    kmp_total_comparisons = 0
    rk_total_comparisons = 0

    for _ in range(num_iterations):
        _, c1 = naive_search(text_large, p)
        _, c2 = kmp_search(text_large, p)
        _, c3 = rabin_karp(text_large, p)

        naive_total_comparisons += c1
        kmp_total_comparisons += c2
        rk_total_comparisons += c3

    naive_avg_comparisons = naive_total_comparisons / num_iterations
    kmp_avg_comparisons = kmp_total_comparisons / num_iterations
    rk_avg_comparisons = rk_total_comparisons / num_iterations

    print(f'{p:>12} {naive_avg_comparisons:>15.2f} {kmp_avg_comparisons:>13.2f} {rk_avg_comparisons:>13.2f}')
```
